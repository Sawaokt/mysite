```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# 1. 천재 AI 뇌 불러오기 (ImageNet이라는 거대한 데이터셋으로 이미 학습된 MobileNetV2 모델)
model = tf.keras.applications.MobileNetV2(weights='imagenet')

# 2. 테스트할 사진 준비하기 (인터넷에 있는 강아지 사진 URL)
image_url = 'https://storage.googleapis.com/download.tensorflow.org/example_images/YellowLabradorLooking_new.jpg'
image_path = tf.keras.utils.get_file('dog.jpg', image_url)

# 3. 사진을 AI가 소화할 수 있는 숫자 배열로 변환하기 (전처리)
img = tf.keras.preprocessing.image.load_img(image_path, target_size=(224, 224)) # 모델이 좋아하는 224x224 사이즈로 조절
img_array = tf.keras.preprocessing.image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0) # AI는 여러 장을 동시에 처리하도록 설계되어 있어, 1장짜리 묶음으로 포장해줌
img_array = tf.keras.applications.mobilenet_v2.preprocess_input(img_array)

# 4. AI에게 사진 보여주며 무엇인지 예측(Predict)하게 하기
print(" ----------loading-------------")
predictions = model.predict(img_array)

# 5. AI의 답변을 사람이 알아볼 수 있는 글자로 번역하기 (상위 3개 결과 출력)
decoded_predictions = tf.keras.applications.mobilenet_v2.decode_predictions(predictions, top=3)[0]

# --- 결과 출력 ---
plt.imshow(img)
plt.axis('off')
plt.show()

print("\n🎉 [AI (result)]")
for i, (class_id, label, probability) in enumerate(decoded_predictions):
    print(f"{i+1}위: {label} (Pr(x): {probability * 100:.2f}%)")


```

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# (이전 셀에서 model은 이미 다운로드했다고 가정합니다)

print("아래의 '파일 선택' 버튼을 눌러 내 컴퓨터/폰에 있는 사진을 업로드하세요!")
uploaded = files.upload()

# 업로드한 파일 이름 가져오기
for filename in uploaded.keys():
    print(f"\n✅ '{filename}' has been uploaded.")

    # 1. 업로드한 내 사진 불러오기 및 사이즈 맞추기
    img = tf.keras.preprocessing.image.load_img(filename, target_size=(224, 224))

    # 2. AI가 읽을 수 있게 숫자 배열로 변환
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = tf.keras.applications.mobilenet_v2.preprocess_input(img_array)

    # 3. AI의 예측!
    print("AI analyzing...")
    predictions = model.predict(img_array)
    decoded_predictions = tf.keras.applications.mobilenet_v2.decode_predictions(predictions, top=3)[0]

    # 4. 내 사진과 결과 출력
    plt.imshow(img)
    plt.axis('off')
    plt.show()

    print("\n🎉 [AI의 분석 결과입니다!]")
    for i, (class_id, label, probability) in enumerate(decoded_predictions):
        print(f"{i+1}위: {label} (확률: {probability * 100:.2f}%)")
```
